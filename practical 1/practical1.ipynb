```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pandas.plotting import parallel_coordinates, radviz

# For inline plotting in Jupyter
%matplotlib inline

# Seaborn style
sns.set(style="white", color_codes=True)
```

```python
# Load the iris dataset
df = pd.read_csv("./iris.csv")  # Make sure iris.csv exists in the current directory
df.head()
```

```python
df.plot(kind="scatter", x="sepal.length", y="sepal.width", color='purple')
plt.title("Pandas Scatter Plot")
plt.show()
```

```python
sns.FacetGrid(df, hue="variety", height=5) \
   .map(plt.scatter, "sepal.length", "sepal.width") \
   .add_legend()
plt.title("FacetGrid - SepalLength vs SepalWidth")
plt.show()
```

```python
sns.jointplot(x="sepal.length", y="sepal.width", data=df, height=5)
plt.show()
```

```python
sns.boxplot(x="variety", y="petal.length", data=df)
plt.title("Boxplot - PetalLengthCm by Species")
plt.show()
```

```python
ax = sns.boxplot(x="variety", y="petal.length", data=df)
sns.stripplot(x="variety", y="petal.length", data=df, jitter=True, edgecolor="gray")
plt.title("Box + Strip Plot - PetalLengthCm by Species")
plt.show()
```

```python
sns.FacetGrid(df, hue="variety", height=6) \
   .map(sns.kdeplot, "petal.length") \
   .add_legend()
plt.title("KDE Plot - PetalLengthCm by Species")
plt.show()
```

```python
sns.violinplot(x="variety", y="petal.length", data=df)
plt.title("Violin Plot - PetalLengthCm by Species")
plt.show()
```

```python
sns.pairplot(df, hue="variety", diag_kind="kde", height=3)
plt.show()
```

```python
df.boxplot(by="variety", figsize=(12, 6))
plt.suptitle("Boxplot of All Features by Variety")
plt.show()
```

```python
parallel_coordinates(df, "variety")
plt.title("Parallel Coordinates Plot")
plt.show()
```
